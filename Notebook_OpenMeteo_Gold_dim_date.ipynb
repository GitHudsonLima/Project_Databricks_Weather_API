```sql

-- Create or replace the date dimension table.
CREATE OR REPLACE TABLE weather_openmeteo.gold.dim_date AS
WITH date_sequence AS (
    SELECT sequence(to_date( '2026-01-01' ), to_date( '2035-12-31' ), INTERVAL 1 DAY) AS date_value
)
SELECT 
    CAST(DATE_FORMAT(d.date_value, 'yyyyMMdd') AS BIGINT) AS date_key,
    d.date_value AS full_date,
    YEAR(d.date_value) AS calendar_year,
    QUARTER(d.date_value) AS calendar_quarter,
    MONTH(d.date_value) AS calendar_month,
    DATE_FORMAT(d.date_value, 'MMMM') AS month_name,
    DAY(d.date_value) AS day_of_month,
    DAYOFWEEK(d.date_value) AS day_of_week,
    DATE_FORMAT(d.date_value, 'EEEE') AS day_name,
    WEEKOFYEAR(d.date_value) AS week_of_year,
    -- Useful Indicators
    CASE WHEN DAYOFWEEK(d.date_value) IN (1, 7) THEN TRUE ELSE FALSE END AS is_weekend,
    -- Holidays
    CASE WHEN DATE_FORMAT(d.date_value, 'MM-dd') IN ('01-01', '5-25', '12-25') THEN TRUE ELSE FALSE END AS is_holiday
FROM date_sequence
LATERAL VIEW EXPLODE(date_value) d AS date_value;

Select * from weather_openmeteo.gold.dim_date
```
